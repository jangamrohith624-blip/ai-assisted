```python
import pandas as pd
df=pd.read_csv("/content/StudentsPerformance (1).csv")
df
```

```python
import numpy as np

math_scores = df['math score'].to_numpy()
science_scores = df['writing score'].to_numpy() # Assuming 'science_score' was intended to be 'writing score'
english_scores = df['reading score'].to_numpy()

print(f"Math Scores (NumPy array, first 5 elements):\n{math_scores[:5]}\n")
print(f"Science Scores (NumPy array, first 5 elements):\n{science_scores[:5]}\n")
print(f"English Scores (NumPy array, first 5 elements):\n{english_scores[:5]}\n")
```

```python
print("--- Math Score Statistics ---")
print(f"Mean Math Score: {np.mean(math_scores):.2f}")
print(f"Median Math Score: {np.median(math_scores):.2f}")
print(f"Standard Deviation Math Score: {np.std(math_scores):.2f}\n")

print("--- Science Score Statistics ---")
print(f"Mean Science Score: {np.mean(science_scores):.2f}")
print(f"Median Science Score: {np.median(science_scores):.2f}")
print(f"Standard Deviation Science Score: {np.std(science_scores):.2f}\n")

print("--- English Score Statistics ---")
print(f"Mean English Score: {np.mean(english_scores):.2f}")
print(f"Median English Score: {np.median(english_scores):.2f}")
print(f"Standard Deviation English Score: {np.std(english_scores):.2f}\n")
```

```python
print("--- Math Score Min/Max ---")
print(f"Minimum Math Score: {np.min(math_scores)}")
print(f"Maximum Math Score: {np.max(math_scores)}\n")

print("--- Science Score Min/Max ---")
print(f"Minimum Science Score: {np.min(science_scores)}")
print(f"Maximum Science Score: {np.max(science_scores)}\n")

print("--- English Score Min/Max ---")
print(f"Minimum English Score: {np.min(english_scores)}")
print(f"Maximum English Score: {np.max(english_scores)}\n")
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))
plt.hist(math_scores, bins=20, color='lightgreen', edgecolor='black')
plt.title('Distribution of Math Scores (without Seaborn)')
plt.xlabel('Math Score')
plt.ylabel('Frequency')
plt.grid(axis='y', alpha=0.75)
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Calculate average scores
avg_math = np.mean(math_scores)
avg_science = np.mean(science_scores)
avg_english = np.mean(english_scores)

subjects = ['Math', 'Science', 'English']
average_scores = [avg_math, avg_science, avg_english]

plt.figure(figsize=(8, 6))
plt.bar(subjects, average_scores, color=['skyblue', 'lightcoral', 'lightgreen'])
plt.title('Average Scores by Subject')
plt.xlabel('Subject')
plt.ylabel('Average Score')
plt.grid(axis='y', alpha=0.75)

plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 7))
sns.scatterplot(x=math_scores, y=english_scores, alpha=0.5)
plt.title('Scatter Plot of Math Scores vs. English Scores')
plt.xlabel('Math Score')
plt.ylabel('English Score')
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```

```python
gender_scores = df.groupby('gender')[['math score', 'writing score', 'reading score']].mean()
display(gender_scores)
```

```python
parent_education_scores = df.groupby('parental level of education')[['math score', 'writing score', 'reading score']].mean()
display(parent_education_scores)
```

```python
import matplotlib.pyplot as plt

# Plotting gender-wise average scores
gender_scores.plot(kind='bar', figsize=(10, 6), colormap='viridis')
plt.title('Average Scores by Gender')
plt.xlabel('Gender')
plt.ylabel('Average Score')
plt.xticks(rotation=0)
plt.grid(axis='y', alpha=0.75)
plt.legend(title='Subject')
plt.tight_layout()
plt.show()
```
